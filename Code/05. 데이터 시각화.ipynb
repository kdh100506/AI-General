```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_excel('../Data/05. dust_pm25.xlsx', index_col='area')
plt.figure(figsize=(15,3))

for year in range(2019,2022):
  y_data=data[year]
  plt.plot(y_data, marker = 'o', label = year)

plt.xlabel('area')
plt.ylabel('micrometer')
plt.xticks(rotation = 45)
plt.title('2019~2021 Dust Line-Graph')
plt.legend()
plt.show()

data
```

```python
data2017 = data[2017]
plt.hist(data2017, bins=9, label='2.5pm')
plt.xlabel('pm2.5')
plt.ylabel('frequency')
plt.title('2017 Dust Histogram')
plt.legend()
plt.grid()
plt.show()
```
