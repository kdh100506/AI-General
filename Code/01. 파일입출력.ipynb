```python
# 입력받아 쓰기 (write)

file = open('../Data/01. name.txt', 'w', encoding='utf-8')
while True:
  team = input('팀 : ')
  if not team:
    print('프로그램이 종료 됩니다.')
    break
  file.write(team + '\n')
file.close()
```

```python
# 입력받아 쓰기 (append)

file = open('../Data/01. name.txt', 'a', encoding='utf-8')
while True:
  team = input('팀 : ')
  if not team:
    print('프로그램이 종료 됩니다.')
    break
  file.write(team + '\n')
file.close()
```
