```python
# 입력받아 쓰기 (write)

file = open('../Data/01. name.txt', 'w', encoding='utf-8')
while True:
  team = input('팀 : ')
  if not team:
    print('프로그램이 종료 됩니다.')
    break
  file.write(team + '\n')
file.close()
```

```python
# 입력받아 쓰기 (append)

file = open('../Data/01. name.txt', 'a', encoding='utf-8')
while True:
  team = input('팀 : ')
  if not team:
    print('프로그램이 종료 됩니다.')
    break
  file.write(team + '\n')
file.close()
```

```python
# 새로운 정보 추가

import pandas as pd

df = pd.read_excel('../Data/01. club.xlsx')
df.head()
add_df = pd.DataFrame([{
  '학번' : 10512, 
  '이름' : '김하늘', 
  '학년' : 1, 
  '동아리' : 'AI 연구부', 
  '역할' : ' 부원'
  }])

new_df = pd.concat([df, add_df], ignore_index=True) #ignore_index=True 는 인덱스를 순서대로 정렬
new_df.to_excel('../Data/01. new_club.xlsx', index=False)
new_df.tail()
```
