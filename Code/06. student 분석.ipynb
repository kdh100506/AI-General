```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../Data/06. student.csv')
df
```

```python
plt.hist(df['StudyHours'])
plt.title('Study Hours Distribution')
plt.xlabel('Study Hours')
plt.ylabel('Freqquency')
plt.show()
```

```python
plt.scatter(x = df['StudyHours'], y = df['AssignmentScore'])
plt.xlabel('Study Hours')
plt.ylabel('Assignment Score')
plt.title('Study Hours VS Assignment Score')
```

```python
# 통과와 미통과 집단별 평균 구한 후, 공부시간 비교

result_mean = df.groupby('Result')['StudyHours'].mean()
```

```python
#두 집단을 막대 그래프로 나타내기

plt.bar(result_mean.index, result_mean.values)
plt.xlabel('Result')
plt.xticks([0,1])
plt.ylabel('Average Study Hours')
plt.title('Average Study Hours by Result')
plt.show()
```

```python
plt.scatter(x = df['Attendance'], y = df['AssignmentScore'])
plt.xlabel('Attendance')
plt.ylabel('Assignment Score')
plt.title('Attendance and Assignment Score')
plt.show()
```

```python
plt.hist(x = df['SleepHours'], bins=5)
plt.xlabel('Sleep Hours')
plt.ylabel('Freqquency')
plt.title('Sleep Hours Distribution')
```

```python
result_mean = df.groupby('Result')['SleepHours'].mean()

plt.bar(result_mean.index, result_mean.values)
plt.xlabel('Result')
plt.xticks([0,1])
plt.ylabel('Average Sleep Hours')
plt.title('Average Sleep Hours by Result')
plt.show()
```
